# ARDL bounds and diagnostics

Six baseline equations: four 31-day exposure-tier shares and two log engagement rates. Each equation includes platform-labeled supply and unlabeled supply. The table follows the manuscript diagnostics: bounds F, the Newey-West t on rho, Breusch-Godfrey p-values at lags 1 and 7, and the CUSUM and CUSUMSQ ratios. A ratio below 1 means the recursive-residual path stays inside the 5% band. The last cell is the short lag grid.

### Baseline equations

```python
from pathlib import Path
import sys

here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    if (candidate / "run_pipeline.py").exists():
        sys.path.insert(0, str(candidate))
        root = candidate
        break
else:
    raise FileNotFoundError("Open this notebook from the repository.")

import pandas as pd
from core.pipeline import load_config, run_engagement, run_exposure_tiers

cfg = load_config(root)
daily = pd.read_csv(root / "data" / "synthetic" / "daily_panel.csv")
tiers = run_exposure_tiers(root, cfg, daily)
engage = run_engagement(root, cfg, daily)
table = pd.concat([tiers, engage], ignore_index=True)
table = table[table["regressor"].isin(["labeled_supply", "log_labeled_supply"])]
keep = [
    "outcome", "p", "q", "rho", "t_rho", "t_bound", "bounds_f", "bounds_i0", "bounds_i1",
    "bounds_decision", "theta", "theta_ci95_low", "theta_ci95_high",
    "breusch_godfrey_p_lag1", "breusch_godfrey_p_lag7", "cusum_ratio", "cusumsq_ratio",
]
table[keep]
```
