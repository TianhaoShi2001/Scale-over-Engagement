```python
from pathlib import Path
import sys

here = Path.cwd().resolve()
for candidate in [here, *here.parents]:
    if (candidate / "run_pipeline.py").exists():
        sys.path.insert(0, str(candidate))
        root = candidate
        break
else:
    raise FileNotFoundError("Open this notebook from the repository.")

import pandas as pd
from core.ardl import ardl_equation
from core.pipeline import load_config

cfg = load_config(root)
daily = pd.read_csv(root / "data" / "synthetic" / "daily_panel.csv")
fit = ardl_equation(
    daily,
    "rel_exposure",
    ["rel_supply", "rel_engagement"],
    cfg["time_exog"],
    p=2,
    q=2,
    hac_lags=cfg["hac_lags"],
)
fit.to_csv(root / "data" / "results" / "level_ardl.csv", index=False)
fit
```
